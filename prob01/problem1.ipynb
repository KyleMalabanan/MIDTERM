```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Define the transfer function H2(s) = (s + 1) / (s^4 + 4s^3 + 7s^2 + 6s + 3)
num = [1, 1]
den = [1, 4, 7, 6, 3]
system = signal.TransferFunction(num, den)

# Calculate and plot the step response
t, y = signal.step(system)

plt.figure(figsize=(8, 5))
plt.plot(t, y)
plt.title('Step Response of H2(s)')
plt.xlabel('Time (s)')
plt.ylabel('Displacement X2')
plt.grid(True)
plt.savefig('step_response.png')
plt.close()

# Extract transient behavior metrics
peak_response = np.max(y)
final_value = num[-1] / den[-1]
overshoot = (peak_response - final_value) / final_value * 100

# Calculate settling time (2% tolerance)
tolerance = 0.02 * final_value
indices_outside = np.where(np.abs(y - final_value) >= tolerance)[0]
settling_time = t[indices_outside[-1]] if len(indices_outside) > 0 else 0

# Generate a pole-zero map
poles = np.roots(den)
zeros = np.roots(num)

plt.figure(figsize=(8, 5))
plt.scatter(np.real(poles), np.imag(poles), marker='x', color='red', label='Poles')
plt.scatter(np.real(zeros), np.imag(zeros), marker='o', color='blue', label='Zeros')
plt.axvline(0, color='black', lw=1)
plt.axhline(0, color='black', lw=1)
plt.title('Pole-Zero Map of H2(s)')
plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.grid(True)
plt.legend()
plt.savefig('pole_zero_map.png')
plt.close()
```
