```python
import numpy as np
import scipy.signal as signal
import matplotlib.pyplot as plt

# Parameters
M1 = M2 = B1 = B2 = B3 = K1 = K2 = K3 = 1.0

# Transfer Function Coefficients
num = [B3, K2]
den = [
    M1 * M2,
    (B1 * M2 + B2 * M1 + B3 * M1 + B3 * M2),
    (B1 * B2 + B1 * B3 + B2 * B3 + K1 * M2 + K2 * M1 + K2 * M2 + K3 * M1),
    (B1 * K2 + B1 * K3 + B2 * K1 + B2 * K2 + B3 * K1 + B3 * K3),
    (K1 * K2 + K1 * K3 + K2 * K3)
]

# 1. Define the system
sys = signal.TransferFunction(num, den)

# 2. Generate Step Response
t, y = signal.step(sys, T=np.linspace(0, 20, 1000))

# 3. Steady-state value, Overshoot, and Settling Time
steady_state = num[-1] / den[-1]
peak_value = np.max(y)
overshoot_pct = max(0, (peak_value - steady_state) / steady_state * 100)

# Settling time (2% criterion)
threshold = 0.02 * steady_state
settling_time = 0
for i in range(len(y) - 1, -1, -1):
    if abs(y[i] - steady_state) > threshold:
        settling_time = t[i]
        break

# 4. Poles and Zeros
zeros = sys.zeros
poles = sys.poles

# 5. Stability
is_stable = all(p.real < 0 for p in poles)

# Plotting output generated as image media...
```
